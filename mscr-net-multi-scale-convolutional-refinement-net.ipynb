```python
# ====================== IMPORTS ======================
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# ====================== PARAMETERS ======================
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50

# Dataset Path
DATASET_PATH = "/kaggle/input/datasets/anandjain1112/papaya-sugarcane-9026/Papaya-sugarcane dataset"

# ====================== LOAD DATASET ======================
train_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

test_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

# ====================== CLASS NAMES ======================
class_names = train_dataset.class_names
NUM_CLASSES = len(class_names)

print("Classes:", class_names)
print("Number of Classes:", NUM_CLASSES)

# ====================== PREFETCH ======================
AUTOTUNE = tf.data.AUTOTUNE

train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
test_dataset = test_dataset.prefetch(buffer_size=AUTOTUNE)

# ====================== DATA AUGMENTATION ======================
data_augmentation = models.Sequential([
    layers.RandomFlip("horizontal")
    
])

# =========================================================
# CHANNEL ATTENTION BLOCK
# =========================================================




def lightweight_attention(x, reduction=16):

    channels = x.shape[-1]

    # Depthwise Convolution
    conv = layers.DepthwiseConv2D(
        kernel_size=3,
        padding='same',
        activation='relu'
    )(x)

    conv = layers.BatchNormalization()(conv)

    # Global Average Pooling
    gap = layers.GlobalAveragePooling2D()(conv)

    # Dense Layer
    dense1 = layers.Dense(
        channels // reduction,
        activation='swish'
    )(gap)

    dense2 = layers.Dense(
        channels,
        activation='hard_sigmoid'
    )(dense1)

    # Reshape
    scale = layers.Reshape((1,1,channels))(dense2)

    # Attention Scaling
    attention = layers.Multiply()([x, scale])

    # Final Depthwise Convolution
    output = layers.DepthwiseConv2D(
        kernel_size=3,
        padding='same',
        activation='relu'
    )(attention)

    output = layers.BatchNormalization()(output)

    return output

# =========================================================
# INVERTED RESIDUAL BLOCK
# =========================================================
def inverted_residual_block(x, expand, filters, stride):

    input_channels = x.shape[-1]

    shortcut = x

    # =====================================================
    # EXPANSION LAYER
    # =====================================================
    expanded_channels = input_channels * expand

    x = layers.Conv2D(
        expanded_channels,
        kernel_size=1,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.ReLU(max_value=6)(x)

    # =====================================================
    # DEPTHWISE CONVOLUTION
    # =====================================================
    x = layers.DepthwiseConv2D(
        kernel_size=3,
        strides=stride,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.ReLU(max_value=6)(x)

    # =====================================================
    # PROJECTION LAYER
    # =====================================================
    x = layers.Conv2D(
        filters,
        kernel_size=1,
        padding='same',
        use_bias=False
    )(x)

    x = layers.BatchNormalization()(x)

    # =====================================================
    # RESIDUAL CONNECTION
    # =====================================================
    if stride == 1 and input_channels == filters:
        x = layers.Add()([shortcut, x])

    return x

# =========================================================
# INPUT LAYER
# =========================================================
inputs = layers.Input(shape=(224,224,3))

# =========================================================
# DATA AUGMENTATION
# =========================================================
x = data_augmentation(inputs)

# =========================================================
# RESCALING
# =========================================================
x = layers.Rescaling(1./255)(x)

# =========================================================
# INITIAL CONVOLUTION
# =========================================================
x = layers.Conv2D(
    32,
    kernel_size=3,
    strides=2,
    padding='same',
    use_bias=False
)(x)

x = layers.BatchNormalization()(x)

x = layers.ReLU(max_value=6)(x)

# =========================================================
# MOBILENETV2 BLOCKS
# =========================================================

# Block 1
x = inverted_residual_block(x, expand=1, filters=16, stride=1)

# Block 2
x = inverted_residual_block(x, expand=6, filters=24, stride=2)
x = inverted_residual_block(x, expand=6, filters=24, stride=1)

# Block 3
x = inverted_residual_block(x, expand=6, filters=32, stride=2)
x = inverted_residual_block(x, expand=6, filters=32, stride=1)
x = inverted_residual_block(x, expand=6, filters=32, stride=1)

# Block 4
x = inverted_residual_block(x, expand=6, filters=64, stride=2)
x = inverted_residual_block(x, expand=6, filters=64, stride=1)
x = inverted_residual_block(x, expand=6, filters=64, stride=1)
x = inverted_residual_block(x, expand=6, filters=64, stride=1)

# Block 5
x = inverted_residual_block(x, expand=6, filters=96, stride=1)
x = inverted_residual_block(x, expand=6, filters=96, stride=1)
x = inverted_residual_block(x, expand=6, filters=96, stride=1)


# Block 6
x = inverted_residual_block(x, expand=6, filters=160, stride=2)
x = inverted_residual_block(x, expand=6, filters=160, stride=1)
x = inverted_residual_block(x, expand=6, filters=160, stride=1)


x = inverted_residual_block(x, expand=6, filters=192, stride=2)
x = inverted_residual_block(x, expand=6, filters=192, stride=1)
x = inverted_residual_block(x, expand=6, filters=192, stride=1)
# Block 7
x = inverted_residual_block(x, expand=6, filters=320, stride=1)

# =========================================================
# FINAL CONVOLUTION LAYER
# =========================================================
x = layers.Conv2D(
    1280,
    kernel_size=1,
    padding='same',
    use_bias=False
)(x)

x = layers.BatchNormalization()(x)

x = layers.ReLU(max_value=6)(x)

# =========================================================
# CHANNEL ATTENTION AFTER FULL MODEL
# =========================================================
x = lightweight_attention(x)
# =========================================================
# GLOBAL AVERAGE POOLING
# =========================================================
x = layers.GlobalAveragePooling2D()(x)

# =========================================================
# DROPOUT
# =========================================================
x = layers.Dropout(0.8)(x)

# =========================================================
# DENSE LAYER
# =========================================================
x = layers.Dense(256, activation='relu')(x)

# =========================================================
# OUTPUT LAYER
# =========================================================
outputs = layers.Dense(
    NUM_CLASSES,
    activation='softmax'
)(x)

# =========================================================
# FINAL MODEL
# =========================================================
model = models.Model(inputs, outputs)

# ====================== MODEL SUMMARY ======================
model.summary()

# ====================== COMPILE MODEL ======================
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ====================== TRAIN MODEL ======================
history = model.fit(
    train_dataset,
    validation_data=test_dataset,
    epochs=EPOCHS
)

# ====================== EVALUATE MODEL ======================
test_loss, test_accuracy = model.evaluate(test_dataset)

print("\nTest Accuracy:", test_accuracy)
print("Test Loss:", test_loss)

# ====================== ACCURACY GRAPH ======================
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Accuracy Graph')
plt.legend()

# ====================== LOSS GRAPH ======================
plt.subplot(1,2,2)

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss Graph')
plt.legend()

plt.show()

# ====================== PREDICTIONS ======================
y_true = []
y_pred = []

for images, labels in test_dataset:

    predictions = model.predict(images)

    predicted_labels = np.argmax(predictions, axis=1)

    y_true.extend(labels.numpy())
    y_pred.extend(predicted_labels)

# ====================== CONFUSION MATRIX ======================
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(15,15))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()

# ====================== CLASSIFICATION REPORT ======================
print("\nClassification Report:\n")

print(classification_report(
    y_true,
    y_pred,
    target_names=class_names
))

# ====================== SAVE MODEL ======================
model.save("MobileNetV2_ChannelAttention_AfterModel.h5")

print("\nModel Saved Successfully")
```
